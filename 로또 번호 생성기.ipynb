```python
"""로또 번호 생성기 : 1~45 중 중복 없는 6개 번호를 생성합니다."""
import random


def generate_lotto() -> list[int]:
    """random.sample은 중복 없이 뽑아주므로 정렬만 하면 된다."""
    return sorted(random.sample(range(1, 46), 6))


def main() -> None:
    raw = input("몇 게임을 생성할까요? (기본 1): ").strip()
    games = int(raw) if raw.isdigit() and int(raw) > 0 else 1

    for i in range(1, games + 1):
        numbers = generate_lotto()
        print(f"{i}게임: {' '.join(f'{n:02d}' for n in numbers)}")


if __name__ == "__main__":
    main()
```
