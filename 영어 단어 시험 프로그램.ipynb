```python
"""영어 단어 시험 프로그램 : 영어 단어의 한국어 뜻을 맞혀보세요."""
import random

WORDS = {
    "apple": "사과",
    "book": "책",
    "computer": "컴퓨터",
    "school": "학교",
    "friend": "친구",
    "water": "물",
    "happy": "행복한",
    "study": "공부하다",
}


def run_quiz(words: dict[str, str]) -> None:
    items = list(words.items())
    random.shuffle(items)
    score = 0

    for english, korean in items:
        answer = input(f"'{english}'의 뜻은? ").strip()
        if answer == korean:
            print("정답!\n")
            score += 1
        else:
            print(f"오답! 정답은 '{korean}'입니다.\n")

    print(f"결과: {len(items)}문제 중 {score}문제 정답")


if __name__ == "__main__":
    run_quiz(WORDS)
```
