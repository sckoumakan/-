```python
"""숫자 맞히기 게임 : 1~100 사이의 랜덤 숫자를 맞춰보세요."""
import random


def play() -> None:
    answer = random.randint(1, 100)
    tries = 0
    print("1부터 100 사이의 숫자를 맞혀보세요!")

    while True:
        raw = input("숫자 입력: ").strip()
        if not raw.isdigit():
            print("숫자만 입력해주세요.")
            continue

        guess = int(raw)
        if not 1 <= guess <= 100:
            print("1~100 사이의 숫자를 입력해주세요.")
            continue

        tries += 1
        if guess < answer:
            print("UP! 더 큰 숫자입니다.")
        elif guess > answer:
            print("DOWN! 더 작은 숫자입니다.")
        else:
            print(f"정답입니다! {tries}번 만에 맞혔어요.")
            break


if __name__ == "__main__":
    play()
```
