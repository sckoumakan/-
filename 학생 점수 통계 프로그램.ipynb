```python
"""학생 점수 통계 프로그램 : 평균, 최고점, 최저점을 계산합니다."""


def read_scores() -> list[float]:
    """점수를 한 줄씩 입력받고, 빈 줄을 입력하면 종료한다."""
    scores: list[float] = []
    print("점수를 입력하세요. (입력을 마치려면 빈 줄에서 Enter)")

    while True:
        raw = input(f"{len(scores) + 1}번 학생 점수: ").strip()
        if raw == "":
            break
        try:
            score = float(raw)
        except ValueError:
            print("숫자로 입력해주세요.")
            continue
        if not 0 <= score <= 100:
            print("0~100 사이의 점수를 입력해주세요.")
            continue
        scores.append(score)

    return scores


def main() -> None:
    scores = read_scores()
    if not scores:
        print("입력된 점수가 없습니다.")
        return

    print("\n[통계]")
    print(f"학생 수: {len(scores)}명")
    print(f"평균: {sum(scores) / len(scores):.2f}")
    print(f"최고점: {max(scores):g}")
    print(f"최저점: {min(scores):g}")


if __name__ == "__main__":
    main()
```
